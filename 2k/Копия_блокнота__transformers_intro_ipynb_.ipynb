```python
# Импорт класса AutoTokenizer для автоматической загрузки токенизатора
from transformers import AutoTokenizer
```

```python
# Загрузка предварительно обученного токенизатора BERT с учетом регистра (cased)
tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")
```

```python
# Определяем входную последовательность (предложение)
sequence = "A Titan RTX has 24GB of VRAM"
# Токенизируем последовательность с использованием загруженного токенизатора
tokenized_sequence = tokenizer.tokenize(sequence)
# Выводим список токенов
print(tokenized_sequence)
```

```python
# Кодируем последовательность в идентификаторы токенов
# Токенизатор автоматически добавляет специальные токены ([CLS], [SEP])
# и создает маску внимания (attention mask)
inp_ids = tokenizer(sequence)
# Выводим идентификаторы токенов
print(inp_ids['input_ids'])
```

```python
# Определяем батч предложений (список строк)
batch_sentences = ["Hello I'm a single sentence",
                   "And another sentence",
                   "And the very very last one"]
# Кодируем батч предложений.
# padding=True добавляет паддинг, чтобы все последовательности в батче имели одинаковую длину.
encoded_inputs = tokenizer(batch_sentences, padding=True)
# Выводим закодированные входные данные, включая input_ids, token_type_ids и attention_mask
print(encoded_inputs)
```

```python
# Декодируем идентификаторы токенов обратно в строку
# Токенизатор автоматически удаляет специальные токены при декодировании
decoded_sequence = tokenizer.decode(inp_ids['input_ids'])
# Выводим декодированную последовательность
print(decoded_sequence)
```

Вот пример выполнения классификации последовательностей с использованием модели, чтобы определить, являются ли две последовательности перефразированием друг друга. Процесс следующий:

1. Создайте экземпляр токенизатора и модели из имени контрольной точки. Модель идентифицируется как модель BERT и загружает в нее веса, хранящиеся в контрольной точке.

2. Постройте последовательность из двух предложений с правильными разделителями для конкретной модели, идентификаторами типов токенов и масками внимания (которые будут созданы автоматически токенизатором).

3. Пропустите эту последовательность через модель, чтобы она была отнесена к одному из двух доступных классов: 0 (не пересказ) и 1 (пересказ).

4. Вычислите softmax результата, чтобы получить вероятности по классам.

5. Распечатайте результаты.

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch

# Загрузка токенизатора и модели для классификации последовательностей
# Модель "bert-base-cased-finetuned-mrpc" специально обучена для задачи определения парафраза (MRPC - Microsoft Research Paraphrase Corpus)
tokenizer = AutoTokenizer.from_pretrained("bert-base-cased-finetuned-mrpc")
model = AutoModelForSequenceClassification.from_pretrained("bert-base-cased-finetuned-mrpc")

# Определение классов выходных данных (не парафраз, парафраз)
classes = ["not paraphrase", "is paraphrase"]
# Определение входных последовательностей (предложений)
sequence_0 = "The company HuggingFace is based in New York City"
sequence_1 = "Apples are especially bad for your health"
sequence_2 = "HuggingFace's headquarters are situated in Manhattan"

# Токенизация пары предложений (sequence_0 и sequence_2), которые являются парафразами.
# return_tensors="pt" указывает, что результат должен быть в формате PyTorch тензоров.
# Токенизатор автоматически добавляет спец.токены ([CLS], [SEP]) и маски внимания.
paraphrase = tokenizer(sequence_0, sequence_2, return_tensors="pt")
# Токенизация пары предложений (sequence_0 и sequence_1), которые не являются парафразами.
not_paraphrase = tokenizer(sequence_0, sequence_1, return_tensors="pt")

# Пропуск токенизированных входных данных через модель для получения логитов классификации.
# Логиты - это необработанные оценки модели перед функцией активации (например, softmax).
paraphrase_classification_logits = model(**paraphrase).logits
not_paraphrase_classification_logits = model(**not_paraphrase).logits

# Применение функции softmax к логитам для получения вероятностей по классам.
# dim=1 указывает, что softmax применяется по измерению классов.
# .tolist()[0] преобразует тензор вероятностей в список Python для первого элемента батча.
paraphrase_results = torch.softmax(paraphrase_classification_logits, dim=1).tolist()[0]
not_paraphrase_results = torch.softmax(not_paraphrase_classification_logits, dim=1).tolist()[0]

# Вывод результатов для пары, которая должна быть парафразом
print("Результаты для пары, которая должна быть парафразом:")
for i in range(len(classes)):
    # Вывод вероятности для каждого класса, округленной до целого процента
    print(f"{classes[i]}: {int(round(paraphrase_results[i] * 100))}%")
# Вывод результатов для пары, которая не должна быть парафразом
print("\nРезультаты для пары, которая не должна быть парафразом:")
for i in range(len(classes)):
    # Вывод вероятности для каждого класса, округленной до целого процента
    print(f"{classes[i]}: {int(round(not_paraphrase_results[i] * 100))}%")
```

```python
from transformers import pipeline
from pprint import pprint

# Создание пайплайна для задачи языкового моделирования с маскированием (fill-mask)
# Пайплайн заполнит [MASK] токен наиболее вероятными вариантами
nlp = pipeline("fill-mask")

# Применение пайплайна к предложению с маскированным токеном.
# nlp.tokenizer.mask_token получает правильное строковое представление токена маски для используемой модели.
# pprint используется для более красивого вывода списка результатов.
pprint(nlp(f"HuggingFace is creating a {nlp.tokenizer.mask_token} that the community uses to solve NLP tasks."))
```

```python
from transformers import AutoModelForMaskedLM, AutoTokenizer
import torch

# Загрузка токенизатора и модели для языкового моделирования с маскированием
tokenizer = AutoTokenizer.from_pretrained("distilbert-base-cased")
model = AutoModelForMaskedLM.from_pretrained("distilbert-base-cased")
# Определяем последовательность с маскированным токеном
sequence = f"Distilled models are smaller than the models they mimic. Using them instead of the large versions would help {tokenizer.mask_token} our carbon footprint."

# Кодируем последовательность в идентификаторы токенов в формате PyTorch тензоров
input = tokenizer.encode(sequence, return_tensors="pt")
# Находим индекс маскированного токена в последовательности
mask_token_index = torch.where(input == tokenizer.mask_token_id)[0, 1]
# Получаем логиты (необработанные оценки) модели для всех токенов в последовательности
token_logits = model(input).logits
# Выбираем логиты только для маскированного токена
mask_token_logits = token_logits[0, mask_token_index, :]
# Получаем индексы 5 токенов с наибольшими логитами (наиболее вероятные токены для маски)
top_5_tokens = torch.topk(mask_token_logits, 5).indices.numpy()
```

```python
# Декодируем и выводим предложения с заполненной маской для каждого из топ-5 токенов
for token in top_5_tokens:
    print(sequence.replace(tokenizer.mask_token, tokenizer.decode([token])))
```

```python
from transformers import AutoModelForTokenClassification, AutoTokenizer
import torch

# Загрузка модели и токенизатора для задачи распознавания именованных сущностей (NER)
# Модель "dbmdz/bert-large-cased-finetuned-conll03-english" специально обучена для NER на английском языке
model = AutoModelForTokenClassification.from_pretrained("dbmdz/bert-large-cased-finetuned-conll03-english")
tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")
# Определяем входную последовательность
sequence = "Hugging Face Inc. is a company based in New York City. Its headquarters are in DUMBO, therefore very" \
           "close to the Manhattan Bridge."
# Токенизация последовательности, включая специальные токены
# Используется небольшая хитрость для получения токенов со специальными токенами для вывода
tokens = tokenizer.tokenize(tokenizer.decode(tokenizer.encode(sequence)))
# Кодирование последовательности в идентификаторы токенов в формате PyTorch тензоров
inputs = tokenizer.encode(sequence, return_tensors="pt")
# Получение логитов (необработанных оценок) модели для классификации токенов
outputs = model(inputs).logits
# Получение предсказанных классов (индексов) для каждого токена путем выбора индекса с максимальным логитом по измерению классов
predictions = torch.argmax(outputs, dim=2)
```

```python
# Итерация по токенам и их предсказанным меткам сущностей
for token, prediction in zip(tokens, predictions[0].numpy()):
    # Вывод токена и соответствующей метки сущности, полученной из конфигурации модели
    print((token, model.config.id2label[prediction]))
```
